```python
# Import library dan load dataset
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# load dataset
df = pd.read_csv('StudentsPerformance.csv')
df.shape
```

```python
df.head(5)
```

```python
df.info()
```

```python
df.describe()
```

```python
# Visualisasi distribusi nilai
plt.figure(figsize=(8,5))
sns.histplot(
    data=df,
    x = 'math score',
    bins=15,
    kde=True
)
plt.title('distribusi nilai matematika')
plt.xlabel('nilai')
plt.ylabel('jumlah')
plt.show()
```

```python
# Membandingkan nilai rata-rata berdasarkan gender
nilai = [
    'math score',
    'reading score',
    'writing score'
]

rata_rata = df.groupby('gender')[nilai].mean()
rata_rata.plot(
    kind='bar',
    figsize=(9,5)
)
plt.xticks(rotation=15)
plt.title('rata-rata nilai berdasarkan gender')
plt.xlabel('gender')
plt.ylabel('nilai')
plt.show()
```

```python
# Melihat hubungan nilai matematika dan minat baca
plt.figure(figsize=(8,5))
sns.scatterplot(
    data=df,
    x='math score',
    y='reading score',
    hue='gender',
    alpha=0.5

)
plt.show()

```

```python
# Melihat korelasi ketiga nilai ujian

kolom_nilai = [
    'math score',
    'reading score',
    'writing score'
]

korelasi = df[kolom_nilai].corr()

plt.figure(figsize=(8,5))
sns.heatmap(
    data=korelasi,
    annot=True,
    cmap='coolwarm',
    vmin=-1,
)
plt.show()
```

```python
# Membaca outlier (pencilan)

plt.figure(figsize=(8,5))
sns.boxplot(
    data=df[['math score', 'reading score', 'writing score']],
)
plt.show()
```
